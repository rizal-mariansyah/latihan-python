```python
tambah = lambda a, b : a + b

print (tambah(10,10))
```

```python
cek_nilai = lambda n : "lulus" if n >= 75 else "tidak lulus"

print(cek_nilai(80))
print(cek_nilai(60))
```

```python
data = [80, 65, 90, 55, 75, 40, 95]

cek_lulus = filter(lambda n : n >=75 , data)

print(list(cek_lulus))
```

```python
harga = [10000, 20000, 30000, 40000]

harga_pajak = map(lambda x: x * 1.10, harga )


print(list(harga_pajak))
```

```python
nilai = [60, 75, 80, 90, 45, 100]

cek_nilai2 = filter(lambda x : x >= 75 , nilai)
tambah_nilai = map(lambda cek_nilai2 : cek_nilai2 + 5 , cek_nilai2)

print(list(tambah_nilai))
```

```python
harga = [5000, 10000, 15000, 20000, 25000]

before = list(filter (lambda x : x >= 10000 , harga)) 
print(list(before))

after = map (lambda x :x - 2000, before)
print(list(after))


```

```python
penjualan = [5000, 15000, 25000, 8000, 30000, 12000]

harga_min = list(filter (lambda x : x >= 10000, penjualan))

hasil_penjualan = map (lambda x : int(x* 1.10), harga_min)

print(list(harga_min))
print(list(hasil_penjualan))
```
